El corazón de la transferencia. Comparamos la **forma** de la demanda en NYC (donde tenemos ground truth de taxi) con la **movilidad real de Terrassa** (proxy MITMA).

La idea clave: **la escala NO se transfiere** (Manhattan ≠ Terrassa), pero **la forma sí** (cómo varía la demanda con la hora y el día). Aquí lo demostramos visual y cuantitativamente.

Entradas: `perfil_movilidad_terrassa.csv` (de R) y `data/processed/demanda_nyc_2023.parquet` (Fase 1).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

TERRASSA_CSV = 'perfil_movilidad_terrassa.csv'      # ajusta la ruta si hace falta
NYC_PARQUET  = 'data/processed/demanda_nyc_2023.parquet'

ter = pd.read_csv(TERRASSA_CSV)
nyc = pd.read_parquet(NYC_PARQUET)

# Normaliza un perfil a proporción (suma = 1) para comparar FORMAS, no escalas
def norm(s):
    return s / s.sum()

print('Terrassa:', ter.shape, '| NYC:', nyc.shape)
```

## 1. Perfil de movilidad de Terrassa
Mapa de calor distrito × hora (igual que hicimos con NYC). Cada distrito debería tener su propio patrón horario.

```python
piv_ter = ter.pivot_table(index='id_origin', columns='hour', values='viajes', aggfunc='mean')

plt.figure(figsize=(12, 4))
sns.heatmap(piv_ter, cmap='mako_r')
plt.title('Terrassa — movilidad media por distrito × hora')
plt.xlabel('Hora'); plt.ylabel('Distrito (id)')
plt.tight_layout(); plt.show()
```

## 2. Curvas de Terrassa (hora y día de la semana)

```python
labels = ['Lun', 'Mar', 'Mié', 'Jue', 'Vie', 'Sáb', 'Dom']

ter_h = ter.groupby('hour')['viajes'].mean()
ter_w = ter.groupby('dow')['viajes'].mean(); ter_w.index = labels   # dow 1..7 = Lun..Dom

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ter_h.plot(marker='o', ax=ax[0], title='Terrassa — por hora'); ax[0].set_xlabel('Hora')
ter_w.plot(kind='bar', ax=ax[1], color='teal', title='Terrassa — por día'); ax[1].set_xlabel('')
plt.tight_layout(); plt.show()
```

## 3. Perfil de NYC (para comparar)
Calculamos la misma forma a partir de la demanda de taxi de NYC. Ojo: el día de la semana de NYC viene como 0=Domingo … 6=Sábado, así que lo reordenamos a Lun-Dom.

```python
nyc_h = nyc.groupby('hora')['demanda'].mean()

orden_nyc = [1, 2, 3, 4, 5, 6, 0]   # Lun..Sáb, Dom  (dia_semana: 0=Dom)
nyc_w = nyc.groupby('dia_semana')['demanda'].mean().reindex(orden_nyc)
nyc_w.index = labels
```

## 4. ⭐ La comparación: ¿se transfiere la forma?
Superponemos las dos curvas **normalizadas** (proporción del total diario). Si las formas se parecen, la transferencia tiene fundamento.

```python
plt.figure(figsize=(11, 5))
plt.plot(range(24), norm(nyc_h).values, marker='o', label='NYC (taxi)')
plt.plot(range(24), norm(ter_h).values, marker='s', label='Terrassa (movilidad)')
plt.title('Forma horaria NORMALIZADA: NYC vs Terrassa')
plt.xlabel('Hora'); plt.ylabel('Proporción de la actividad diaria')
plt.xticks(range(0, 24, 2)); plt.legend()
plt.tight_layout(); plt.show()

r_hora = np.corrcoef(norm(nyc_h).values, norm(ter_h).values)[0, 1]
print(f'Correlación de la forma horaria (NYC vs Terrassa): r = {r_hora:.3f}')
```

```python
comp_w = pd.DataFrame({'NYC (taxi)': norm(nyc_w), 'Terrassa (movilidad)': norm(ter_w)})

comp_w.plot(kind='bar', figsize=(10, 5))
plt.title('Forma semanal NORMALIZADA: NYC vs Terrassa')
plt.ylabel('Proporción de la actividad semanal'); plt.xlabel('')
plt.xticks(rotation=0)
plt.tight_layout(); plt.show()

r_dia = np.corrcoef(norm(nyc_w).values, norm(ter_w).values)[0, 1]
print(f'Correlación de la forma semanal (NYC vs Terrassa): r = {r_dia:.3f}')
```

**Qué SÍ se transfiere (la forma):**
- Correlación horaria: r = ____ . Si es alta (>0,7), las dos ciudades comparten el ritmo diario (valle de madrugada, actividad de día, pico de tarde) → la forma horaria de NYC es una base válida para Terrassa.
- Correlación semanal: r = ____ .

**Diferencias esperables (a comentar):**

**El argumento de la memoria:**
> Las formas temporales correlacionan a r = ___ : el *patrón* de demanda se transfiere de NYC a Terrassa, mientras que la *escala* (volumen absoluto) se recalibra con la movilidad local. Esto valida la estrategia de transferencia.
